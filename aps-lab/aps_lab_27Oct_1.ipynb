```python
import numpy as np
from statistics import mean, median, mode

# 1. Generate sample data
np.random.seed(42)
data = np.random.randint(10, 101, size=20).tolist()

# 2. Measures of central tendency
data_mean = mean(data)
data_median = median(data)
data_mode = mode(data)

# 3. Measures of dispersion (population formulas)
data_range = max(data) - min(data)
variance = float(np.var(data, ddof=0))
std_dev = float(np.std(data, ddof=0))

q1, q3 = np.percentile(data, [25, 75])
iqr = float(q3 - q1)
quartile_deviation = iqr / 2
mean_abs_deviation = float(np.mean(np.abs(np.array(data) - data_mean)))
cv = (std_dev / data_mean) * 100 if data_mean != 0 else float("nan")

print("Sample data:", data)
print("\nMeasures of Central Tendency")
print(f"Mean:   {data_mean:.2f}")
print(f"Median: {data_median:.2f}")
print(f"Mode:   {data_mode}")

print("\nMeasures of Dispersion")
print(f"Minimum: {min(data)}")
print(f"Maximum: {max(data)}")
print(f"Range: {data_range}")
print(f"Population variance: {variance:.2f}")
print(f"Population standard deviation: {std_dev:.2f}")
print(f"Q1: {q1:.2f}")
print(f"Q3: {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Quartile deviation: {quartile_deviation:.2f}")
print(f"Mean absolute deviation: {mean_abs_deviation:.2f}")
print(f"Coefficient of variation: {cv:.2f}%")
```
