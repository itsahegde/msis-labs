Program to sample from a normal distribution

```python
import numpy as np

rng = np.random.default_rng(seed=42)
pop_mu = 0
pop_sigma = 3
pop_size = 10000

data = rng.normal(loc=pop_mu, scale=pop_sigma, size=pop_size)

print(data[0:10])
```

```python
print("Mean of population:", round(np.mean(data), 2))
print("Standard deviation of population:", round(np.std(data), 2))
```

```python
sample_size = 30
sample = rng.choice(data, size=sample_size, replace=False)
print("Mean of sample:", round(np.mean(sample), 2))
print("Standard deviation of sample:", round(np.std(sample), 2))
print("Standard error of sample mean:", round(np.std(sample) / np.sqrt(sample_size), 2))
```
