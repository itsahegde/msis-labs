```python
from scipy.stats import norm

col1 = df['Oil']

mean1 = np.mean(col1)
stddev1 = np.std(col1)

plt.figure(figsize=(8, 5))
plt.hist(col1, bins=10, density=True, alpha=0.6, color='skyblue', edgecolor='white', label='Oil Data')

x = np.linspace(min(col1), max(col1), 100)
y = norm.pdf(x, mean1, stddev1)

plt.plot(x, y, color='darkblue', linewidth=2, label=f'Normal Fit (mean={mean1:.2f}, stddev={stddev1:.2f})')

plt.title('Normal Distribution Fit to Existing Data', fontsize=14)
plt.xlabel('Values')
plt.ylabel('Density')
plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

x = df.iloc[0:, [1,2,4,5]]
y = df.iloc[0:, 3]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30, random_state=42)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

model = LinearRegression()

model.fit(x_train, y_train)

train_prediction = model.predict(x_train)
test_prediction = model.predict(x_test)

train_mse = mean_squared_error(y_train, train_prediction)
test_mse = mean_squared_error(y_test, test_prediction)
r2score = r2_score(y_test, test_prediction)

print("Train mse", train_mse)
print("Test mse", test_mse)
print("r2 score", r2score)
print()
```
