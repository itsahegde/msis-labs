```python
class Stack:
    def push(self, st, data):
        st.append(data)
    def pop(self, st):
        if len(st)>0:
            ans = st.pop()
            return ans
        else:
            return None
    def top(self, st):
        if len(st)>0:
            ans = st.top()
            return ans
        else:
            return None
    def size(self, st):
        return len(st)
    def isempty(self, st):
        if len(st)>0:
            return False
        else:
            return True
```

```python
st = Stack()
```

```python
s = ['abc', 'def', 'hij', 'lmn', 'opq']
```

```python
st.push(s, "rst")
```

```python
st.size(s)
```

```python
st.isempty(s)
```

```python
st.pop(s)
```
