```python
#Doubly Linked List
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None
    def printfwd(head):
        curr = head
        while curr:
            print(curr.data)
            curr = curr.next
    def gettail(head):
        curr = head
        while curr.next:
            curr = curr.next
        return curr
    def printbwd(head):
        curr = head
        while curr.next:
            curr = curr.next
        while curr:
            print(curr.data)
            curr = curr.prev

```

```python
node1 = Node(5)
node2 = Node(6)
node3 = Node(8)
node4 = Node(9)
node5 = Node(11)

node1.next = node2
node1.prev = None
node2.next = node3
node2.prev = node1
node3.next = node4
node3.prev = node2
node4.next = node5
node4.prev = node3
node5.next = None
node5.prev = node4
```

```python
head = node1
```

```python
Node.printfwd(head)
```

```python
Node.printbwd(head)
```

```python
print(Node.gettail(head).data)
```

```python
#Insertion at end
tail = Node.gettail(head)
node6 = Node(10)
tail.next = node6
node6.prev = tail
node6.next = None
```

```python
Node.printfwd(head)
```

```python
#insertion at start
node7 = Node(12)
head.prev = node7
node7.next = head
node7.prev = None
head = node7
```

```python
Node.printfwd(head)
```

```python
#deletion at last
tail = Node.gettail(head)
tail.prev.next = None
tail.prev = None
tail.next = None

Node.printfwd(head)
```

```python
#deletion at start
head.next.prev = None
head = head.next

Node.printfwd(head)
```

```python
#deletion in middle
val = 9
curr = head
while curr.next.data!=val:
    curr = curr.next
curr.data
```
