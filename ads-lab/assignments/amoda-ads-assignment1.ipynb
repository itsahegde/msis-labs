```python
# 1. Create Single Linked List class with following functionalities

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Display the linked list
    def display(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")

    # 1(a). Add at head
    def add_head(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # 1(b). Add at tail
    def add_tail(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    # 1(c). Delete at head
    def delete_head(self):
        if self.head is None:
            return None

        deleted_data = self.head.data
        self.head = self.head.next
        return deleted_data

    # 1(d). Delete at tail
    def delete_tail(self):
        if self.head is None:
            return None

        if self.head.next is None:
            deleted_data = self.head.data
            self.head = None
            return deleted_data

        current = self.head
        while current.next.next:
            current = current.next

        deleted_data = current.next.data
        current.next = None
        return deleted_data

    # 1(e). Add after given data
    def add_after(self, target, data):
        current = self.head

        while current:
            if current.data == target:
                new_node = Node(data)
                new_node.next = current.next
                current.next = new_node
                return True
            current = current.next

        return False

    # 1(f). Delete node after given data
    def delete_after(self, target):
        current = self.head

        while current:
            if current.data == target:
                if current.next is None:
                    return None

                deleted_data = current.next.data
                current.next = current.next.next
                return deleted_data

            current = current.next

        return None

    # 1(g). Search an element
    def search(self, key):
        current = self.head
        position = 0

        while current:
            if current.data == key:
                return position
            current = current.next
            position += 1

        return -1

    # 2. Find common elements in two linked lists
    @staticmethod
    def common_elements(list1, list2):
        values = set()
        result = LinkedList()

        current = list1.head
        while current:
            values.add(current.data)
            current = current.next

        current = list2.head
        while current:
            if current.data in values:
                result.add_tail(current.data)
            current = current.next

        return result

    # 3. Sum of last n nodes in one traversal
    def sum_last_n(self, n):
        if n < 0:
            raise ValueError("n cannot be negative")

        if n == 0:
            return 0

        fast = self.head
        slow = self.head

        # Move fast n nodes ahead
        for _ in range(n):
            if fast is None:
                raise ValueError("n exceeds list length")
            fast = fast.next

        # Move both pointers until fast reaches the end
        while fast:
            fast = fast.next
            slow = slow.next

        total = 0
        while slow:
            total += slow.data
            slow = slow.next

        return total

    # 4. Reverse the linked list
    def reverse(self):
        previous = None
        current = self.head

        while current:
            next_node = current.next
            current.next = previous
            previous = current
            current = next_node

        self.head = previous

    # 5. Split into alternate elements
    def split(self):
        first = LinkedList()
        second = LinkedList()

        current = self.head
        tail1 = None
        tail2 = None
        take_first = True

        while current:
            next_node = current.next
            current.next = None

            if take_first:
                if first.head is None:
                    first.head = current
                else:
                    tail1.next = current
                tail1 = current
            else:
                if second.head is None:
                    second.head = current
                else:
                    tail2.next = current
                tail2 = current

            take_first = not take_first
            current = next_node

        # Original list is now split into two lists
        self.head = None
        return first, second

    # 6. Check whether linked list is a palindrome
    def is_palindrome(self):
        values = []
        current = self.head

        while current:
            values.append(current.data)
            current = current.next

        return values == values[::-1]

    # 7. Remove duplicate elements efficiently
    def remove_duplicates(self):
        seen = set()
        current = self.head
        previous = None

        while current:
            if current.data in seen:
                previous.next = current.next
            else:
                seen.add(current.data)
                previous = current

            current = current.next

    # 8. Find middle element using slow and fast pointers
    def middle(self):
        if self.head is None:
            return None

        slow = self.head
        fast = self.head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        return slow.data

    # 10. Detect cycle using Floyd's algorithm
    def has_cycle(self):
        slow = self.head
        fast = self.head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

            if slow is fast:
                return True

        return False


# 9. Circular singly linked list
class CircularLinkedList(LinkedList):

    def add_tail(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            new_node.next = new_node
            return

        current = self.head
        while current.next is not self.head:
            current = current.next

        current.next = new_node
        new_node.next = self.head

    def display(self):
        if self.head is None:
            print("Empty list")
            return

        current = self.head
        while True:
            print(current.data, end=" -> ")
            current = current.next
            if current is self.head:
                break
        print("(back to head)")

    # 9. Remove odd elements
    def remove_odd(self):
        if self.head is None:
            return

        # Find the last node
        tail = self.head
        while tail.next is not self.head:
            tail = tail.next

        # Break the circular link temporarily
        tail.next = None

        current = self.head
        previous = None

        while current:
            next_node = current.next

            if current.data % 2 != 0:
                if previous is None:
                    self.head = next_node
                else:
                    previous.next = next_node
            else:
                previous = current

            current = next_node

        # Reconnect the remaining nodes into a circle
        if self.head is None:
            return

        tail = self.head
        while tail.next:
            tail = tail.next

        tail.next = self.head
```
