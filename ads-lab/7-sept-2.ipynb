```python
class FlexiQ:
    def __init__(self, capacity=10):
        self.capacity = capacity
        self.items = [None]*capacity
        self.front = 0
        self.rear = 0
        self.count = 0
    def is_empty(self):
        return self.count == 0
    def is_full(self):
        return self.count == self.capacity
    def resize(self):
        self.capacity*=2
        self.front*=2
    def push(self, data):
        if self.is_full():
            print("Q is full. Resizing now.")
        if self.is_full(): 
            self.resize()
        self.items.append(data)
        self.front+=1
        self.count+=1
    def pop(self):
        if self.is_empty():
            return IndexError("Cannot pop from empty queue")
        item = self.items[self.rear]
        self.rear+=1
        self.count-=1
        return item
    def peek_front(self):
        if self.is_empty():
            return IndexError("Cannot peek from empty queue")
        item = self.items[self.front]
        return item
    def peek_rear(self):
        if self.is_empty():
            return IndexError("Cannot peek from empty queue")
        item = self.items[self.rear]
        return item
    def display(self):
        if self.is_empty():
            return IndexError("Cannot display empty queue")
        curr = self.rear
        while curr<self.front:
            print(curr)
            curr+=1
```

```python
q = FlexiQ(capacity=3)
```

```python
q.display()
```
