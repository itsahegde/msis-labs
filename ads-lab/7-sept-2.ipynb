```python
# Dynamic size queue
```

```python
class FlexiQ:
    def __init__(self, capacity=10):
        self.capacity = capacity
        self.items = [None] * capacity
        self.front = 0
        self.rear = 0
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def is_full(self):
        return self.count == self.capacity

    def resize(self):
        new_capacity = self.capacity * 2
        new_items = [None] * new_capacity
        for i in range(self.count):
            new_items[i] = self.items[(self.front + i) % self.capacity]
        self.items = new_items
        self.front = 0
        self.rear = self.count
        self.capacity = new_capacity

    def push(self, data):
        if self.is_full():
            print("Q is full. Resizing now.")
            self.resize()
        self.items[self.rear] = data
        self.rear = (self.rear + 1) % self.capacity
        self.count += 1

    def pop(self):
        if self.is_empty():
            raise IndexError("Cannot pop from empty queue")
        item = self.items[self.front]
        self.items[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.count -= 1
        return item

    def peek_front(self):
        if self.is_empty():
            raise IndexError("Cannot peek from empty queue")
        return self.items[self.front]

    def peek_rear(self):
        if self.is_empty():
            raise IndexError("Cannot peek from empty queue")
        last_index = (self.rear - 1 + self.capacity) % self.capacity
        return self.items[last_index]

    def display(self):
        if self.is_empty():
            raise IndexError("Cannot display empty queue")
        for i in range(self.count):
            idx = (self.front + i) % self.capacity
            print(self.items[idx])
```

```python
q = FlexiQ(capacity=3)
```
