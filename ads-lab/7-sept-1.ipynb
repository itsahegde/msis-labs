```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = next
class LLStack:
    def __init__(self):
        self.head = None
        self._size = 0
    def is_empty(self):
        return self.head is None
    def push(self, item):
        new_node = Node(item)
        new_node.next = self.head
        self.head = new_node
        self._size+=1
    def pop(self):
        if self.is_empty():
            return IndexError("Cannot pop from empty stack")
        popped_value = self.head.data
        self.head = self.head.next
        self._size-=1
        return popped_value
    def peek(self):
        if self.is_empty():
            return IndexError("Cannot peek from empty stack")
        return self.head.data
    def size(self):
        return self._size
```
