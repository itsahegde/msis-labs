```python
#Linkedlist implementation in python. Using classes.
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
node1 = Node(5)
node2 = Node(6)
node3 = Node(8)
node4 = Node(9)
node5 = Node(11)
```

```python
node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5
node5.next = None
```

```python
head = node1
```

```python
curr = head
```

#Traversal of singly linked list

```python
while curr:
    print(curr.data)
    curr = curr.next
```
