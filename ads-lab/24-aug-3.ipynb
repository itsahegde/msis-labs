```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None
    def printfwd(head, tailnode):
        curr = head
        while curr!=tailnode:
            print(curr.data)
            curr = curr.next
    def gettail(head):
        curr = head
        while curr.next:
            curr = curr.next
        return curr
    def printbwd(tailnode):
        curr = tailnode.prev
        while curr!=head.prev:
            print(curr.data)
            curr = curr.prev
```
