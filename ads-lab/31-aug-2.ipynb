```python
class Q:
    def __init__(self, capacity=100):
        self.front = 0
        self.rear = 0
        self.arr = [None] * capacity

    def enque(self, data):
        if self.rear == len(self.arr):
            print("Queue is full!")
            return
        self.arr[self.rear] = data
        self.rear += 1
        print(f"Element {data} has been enqueued.")

    def deque(self):
        if self.isempty():
            print("Queue is empty!")
            return None
        ans = self.arr[self.front]
        self.front += 1
        print(f"Element {ans} has been dequeued.")
        return ans

    def peekf(self):
        if self.isempty():
            return None
        return self.arr[self.front]

    def peekr(self):
        if self.isempty():
            return None
        return self.arr[self.rear - 1]

    def size(self):
        return self.rear - self.front

    def isempty(self):
        return self.front == self.rear
```

```python
que = Q(3)
```

```python
que.enque(3)
que.enque(4)
```

```python
que.deque()
```

```python
que.enque(5)
```

```python
que.enque(6)
```
