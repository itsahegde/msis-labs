```python
from sklearn.linear_model import Lasso
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
data = {
    'hours': [1, 2, 2, 3, 3, 4, 5, 5, 6, 7, 7, 8, 8, 9, 10],
    'attendence': [55, 60, 65, 62, 70, 72, 75, 80, 78, 82, 85, 88, 90, 92, 95],
    'assignmentscore': [40, 45, 50, 48, 55, 58, 62, 67, 65, 70, 74, 78, 80, 85, 90],
    'studentid': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115],
    'finalmarks': [35, 40, 43, 45, 50, 54, 60, 64, 66, 71, 74, 79, 82, 87, 92]
}
```

```python
df = pd.DataFrame(data)
```

```python
x = df.iloc[:, [0, 1, 2, 3]]
y = df['finalmarks']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=67)
```

```python
model = make_pipeline(
    StandardScaler(),
    Lasso(alpha=0.1)
)

model.fit(x_train, y_train)

y_pred = model.predict(x_test)
```

```python
lasso_model = model.named_steps["lasso"]

coeff_table = pd.DataFrame({
    'Predictor': x.columns,
    'Lasso Coefficient': lasso_model.coef_
})
```

```python
coeff_table
```

```python
print("Using Lasso model: ")
print("MAE: ", mean_absolute_error(y_test, y_pred))
print("MSE: ", mean_squared_error(y_test, y_pred))
print("R2: ", r2_score(y_test, y_pred))
```

```python
lasso_errors = {
    "MAE": [mean_absolute_error(y_test, y_pred)],
    "MSE": [mean_squared_error(y_test, y_pred)],
    "R2": [r2_score(y_test, y_pred)]
}

lasso_error_df = pd.DataFrame(lasso_errors)
```

```python
lasso_error_df
```

Multiple Linear Regression

```python
from sklearn.linear_model import LinearRegression
ml_model = LinearRegression()
ml_model.fit(x_train, y_train)
coeff_table_ml = pd.DataFrame({
    'Feature': x.columns,
    'Coefficients': ml_model.coef_
})
y_pred_ml = ml_model.predict(x_test)

ml_errors = {
    "MAE": [mean_absolute_error(y_test, y_pred_ml)],
    "MSE": [mean_squared_error(y_test, y_pred_ml)],
    "R2": [r2_score(y_test, y_pred_ml)]
}

ml_error_df = pd.DataFrame(ml_errors)
```

Ridge Regression Model

```python
from sklearn.linear_model import Ridge
ridge_model = make_pipeline(
    StandardScaler(),
    Ridge(alpha=1.0)
)
ridge_model.fit(x_train, y_train)
coeff_table_ridge = pd.DataFrame({
    'Feature': x.columns,
    'Coefficients': ridge_model.named_steps['ridge'].coef_
})
y_pred_ridge = ridge_model.predict(x_test)

ridge_errors = {
    "MAE": [mean_absolute_error(y_test, y_pred_ridge)],
    "MSE": [mean_squared_error(y_test, y_pred_ridge)],
    "R2": [r2_score(y_test, y_pred_ridge)]
}

ridge_errors_df = pd.DataFrame(ridge_errors)
```

```python
coeff_table_ridge
```

```python
ridge_errors_df
```

Ridge CV

```python
from sklearn.linear_model import RidgeCV
ridgecv_model = make_pipeline(
    StandardScaler(),
    RidgeCV(alphas=[0.01, 0.1, 1, 10, 100])
)
ridgecv_model.fit(x_train, y_train)
coeff_table_ridgecv = pd.DataFrame({
    'Feature': x.columns,
    'Coefficients': ridgecv_model.named_steps['ridgecv'].coef_
})
y_pred_ridgecv = ridgecv_model.predict(x_test)

ridgecv_errors = {
    "MAE": [mean_absolute_error(y_test, y_pred_ridgecv)],
    "MSE": [mean_squared_error(y_test, y_pred_ridgecv)],
    "R2": [r2_score(y_test, y_pred_ridgecv)]
}

ridgecv_errors_df = pd.DataFrame(ridgecv_errors)
```

```python
coeff_table_ridgecv
```

```python
ridgecv_errors_df
```

Lasso CV

```python
from sklearn.linear_model import LassoCV
lassocv_model = make_pipeline(
    StandardScaler(),
    LassoCV(alphas=[0.01, 0.1, 1, 10, 100])
)
lassocv_model.fit(x_train, y_train)
coeff_table_lassocv = pd.DataFrame({
    'Feature': x.columns,
    'Coefficients': lassocv_model.named_steps['lassocv'].coef_
})
y_pred_lassocv = lassocv_model.predict(x_test)

lassocv_errors = {
    "MAE": [mean_absolute_error(y_test, y_pred_lassocv)],
    "MSE": [mean_squared_error(y_test, y_pred_lassocv)],
    "R2": [r2_score(y_test, y_pred_lassocv)]
}

lassocv_errors_df = pd.DataFrame(lassocv_errors)
```
