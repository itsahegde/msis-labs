```python
# euclidean distance knn implementation
import math
from collections import Counter

def euclidean_distance(point1, point2):
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(point1, point2)))

def knn(train_data, train_labels, test_point, k):
    distances = []

    # step 1: compute the distance between the test point and all training points
    for i in range(len(train_data)):
        dist = euclidean_distance(train_data[i], test_point)
        distances.append((dist, train_labels[i]))

    # step 2: sort by distance
    distances.sort(key=lambda x: x[0])

    #step 3: select k nearest neighbors
    k_nearest_neighbors = distances[:k]

    # extract labels
    labels = [label for _, label in k_nearest_neighbors]

    #step 4: majority vote
    return Counter(labels).most_common(1)[0][0]
```

```python
x_train = [(2,4),(4,4),(4,6),(6,8)]
y_train = ["Red", "Red", "Blue", "Blue"]

test_point = (5, 5)
k=3

print("Predicted class for test point", test_point, "is", knn(x_train, y_train, test_point, k))
```

```python
# ensemble methods boosting adaboost
from sklearn.ensemble import AdaBoostClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv(r"D:\msis-labs\msis-labs\aml-lab\projects\data\mushrooms.csv")
```

```python
data.head(5)
```

```python
y = data['class']
x = data.iloc[:, 1:]

# encode the labels
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
# label encoding for categorical features in x
le_x = LabelEncoder()
for label in x.columns:
    x[label] = le_x.fit_transform(x[label])
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=45)
```

```python
from sklearn.tree import DecisionTreeClassifier
model1 = DecisionTreeClassifier(criterion='entropy', max_depth=1, random_state=42)
model2 = AdaBoostClassifier(estimator = model1, n_estimators=150, learning_rate=0.1, random_state=42)
boostmodel = model2.fit(x_train, y_train)
```

```python
y_preds = boostmodel.predict(x_test)
cm = confusion_matrix(y_test, y_preds)
cm
```

```python
acc = metrics.accuracy_score(y_test, y_preds)
print("Accuracy:", acc*100)
```

```python
# random forest classifier
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier(n_estimators=150, criterion='entropy', random_state=42)
classifier.fit(x_train, y_train)
y_preds = classifier.predict(x_test)
cm = confusion_matrix(y_test, y_preds)
cm
acc = metrics.accuracy_score(y_test, y_preds)
print("Accuracy:", acc*100)
```

```python
# Logistic regression
from sklearn.linear_model import LogisticRegression
logistic_model = LogisticRegression(random_state=42, max_iter=100)
logistic_model.fit(x_train, y_train)
y_preds = logistic_model.predict(x_test)
cm = confusion_matrix(y_test, y_preds)
cm
acc = metrics.accuracy_score(y_test, y_preds)
print("Accuracy:", acc*100)
```

```python
l2_logistic = LogisticRegression(penalty='l2', C=1.0, random_state=42, max_iter=100)
l2_logistic.fit(x_train, y_train)
y_preds = l2_logistic.predict(x_test)
cm = confusion_matrix(y_test, y_preds)
cm
acc = metrics.accuracy_score(y_test, y_preds)
print("Accuracy:", acc*100)
```

```python
l2_logistic = LogisticRegression(penalty='l1', solver='liblinear', C=1.0, random_state=42, max_iter=100)
l2_logistic.fit(x_train, y_train)
y_preds = l2_logistic.predict(x_test)
cm = confusion_matrix(y_test, y_preds)
cm
acc = metrics.accuracy_score(y_test, y_preds)
print("Accuracy:", acc*100)
```

```python
# multi class logistic regression

multiclass_logistic = LogisticRegression(solver='lbfgs', random_state=42, max_iter=1000)
multiclass_logistic.fit(x_train, y_train)
y_preds = multiclass_logistic.predict(x_test)
cm = confusion_matrix(y_test, y_preds)
cm
acc = metrics.accuracy_score(y_test, y_preds)
print("Accuracy:", acc*100)
```
