```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
iris = pd.read_csv("/home/msis/ahegde/aml-lab/projects/data/Iris.csv")
diabetes = pd.read_csv("/home/msis/ahegde/aml-lab/projects/data/diabetes.csv")
social = pd.read_csv("/home/msis/ahegde/aml-lab/projects/data/Social_Network_Ads.csv")
```

```python
datasets = {
    "iris": iris,
    "diabetes": diabetes,
    "social": social
}
```

```python
for name, df in datasets.items():
    print(f"=== EDA FOR: {name} ===")
    print(f"Shape: {df.shape}")
    print(df.head(5))
    print("\n--- Missing Values ---")
    print(df.isnull().sum())
    print("\n--- Duplicated Values ---")
    print(df.duplicated().sum())
    print("\n--- Basic Statistics ---")
    print(df.describe())
    print("\n" + "="*30 + "\n")

```

```python
dataset_info = {
    'iris': {'df': iris, 'bounds': (1, 5)},       # cols 1,2,3,4 -> X | col 5 -> y
    'diabetes': {'df': diabetes, 'bounds': (0, 8)}, # cols 0 to 7 -> X | col 8 -> y
    'social': {'df': social, 'bounds': (2, 4)}    # cols 2,3 -> X | col 4 -> y
}
```

```python
SPLIT_RATIOS = [0.2, 0.25, 0.3, 0.35, 0.4]
CRITERIA = ['gini', 'entropy']
RANDOM_STATE = 42
#STRATIFY=n

results = []

for name, info in dataset_info.items():
    X = info['df'].iloc[:, info['bounds'][0]:info['bounds'][1]].values
    y = info['df'].iloc[:, info['bounds'][1]].values

    for test_size in SPLIT_RATIOS:
    
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=test_size, random_state=RANDOM_STATE, stratify=y
        )
        
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)
        
        for criterion in CRITERIA:
            
            clf = DecisionTreeClassifier(criterion=criterion, random_state=RANDOM_STATE)
            clf.fit(X_train_scaled, y_train)
            
            test_acc = accuracy_score(y_test, clf.predict(X_test_scaled))
            
            results.append({
                'Dataset': name.capitalize(),
                'Test Size': f"{int(test_size * 100)}%",
                'Criterion': criterion,
                'Test Accuracy %': round(test_acc*100, 2)
            })
results_df = pd.DataFrame(results)
#print(results_df.to_string(index=False))
print(results_df)
```
