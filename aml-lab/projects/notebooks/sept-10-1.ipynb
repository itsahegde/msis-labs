```python
import pandas as pd
```

```python
data = {
    "size": [800, 1000, 1200, 1500, 1800],
    "location": ["rural", "urban", "rural", "urban", "urban"],
    "price": [30, 50, 42, 72, 85]
}

df = pd.DataFrame(data)

df_encoded = pd.get_dummies(
    df,
    columns=["location"],
    drop_first=True,
    dtype=int
)

# get_dummies makes a binary matrix of occurence containing each occurence on rows and types on columns
```
