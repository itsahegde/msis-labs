```python
import pandas as pd
```

```python
data = {
    "size": [800, 1000, 1200, 1500, 1800],
    "location": ["rural", "urban", "rural", "urban", "urban"],
    "price": [30, 50, 42, 72, 85]
}

df = pd.DataFrame(data)

df_encoded = pd.get_dummies(
    df,
    columns=["location"],
    drop_first=True,
    dtype=int
)

# get_dummies makes a binary matrix of occurence containing each occurence on rows and types on columns
```

```python
print(df_encoded)
```

```python
# make multiple linear refression for the above dataset
```

```python
x = df_encoded.iloc[:, [0, 2]]
y = df_encoded.iloc[:, [1]]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.pipeline import make_pipeline
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=67)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
ml_model = LinearRegression()
ml_model.fit(x_train, y_train)
# coeff_table_ml = pd.DataFrame({
#     'Feature': x.columns,
#     'Coefficients': ml_model.coef_
# })
y_pred_ml = ml_model.predict(x_test)

ml_errors = {
    "MAE": [mean_absolute_error(y_test, y_pred_ml)],
    "MSE": [mean_squared_error(y_test, y_pred_ml)],
    "R2": [r2_score(y_test, y_pred_ml)]
}

ml_error_df = pd.DataFrame(ml_errors)
```
