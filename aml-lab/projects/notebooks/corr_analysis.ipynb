```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "study_hours": [1, 2, 3,4,5,6],
    "attendence":[50,55,65,70,80,90],
    "previous_marks":[35,40,50,55,65,75],
    "final_marks":[38,45,52,60,70,82]
})
```

```python
corr_mat = data.corr(numeric_only=True)

sns.heatmap(
    corr_mat,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation matrix")
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import MinMaxScaler
# to use chi2 we want non neg numbers, thats why we use minmax scaler

data = load_breast_cancer()
x = data.data
y = data.target
```

```python
print("Number of features: ", x.shape[1])
```

```python
x_nn = MinMaxScaler().fit_transform(x)

selector = SelectKBest(score_func=chi2, k=5)
x_s = selector.fit_transform(x_nn, y)

features_selected = data.feature_names[selector.get_support()]

print("Selected features: ", features_selected)
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
x, y = iris.data, iris.target

print("Number of features: ", x.shape[1])

x_select = SelectKBest(chi2, k=3).fit_transform(x, y)
```

```python
# IRIS ANOVA 

from sklearn.feature_selection import f_classif
import numpy as np

from sklearn.datasets import load_iris

iris = load_iris()
x, y = iris.data, iris.target

selector = SelectKBest(f_classif, k=2)

x_select = selector.fit_transform(x, y)

features_selected = np.array(iris.feature_names)[selector.get_support()]
print("Selected features:", features_selected)
```
