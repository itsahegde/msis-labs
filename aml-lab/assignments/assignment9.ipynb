```python
from sklearn.linear_model import Lasso
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
df1 = pd.read_csv("/home/msis/ahegde/aml-lab/projects/data/data-1.csv")
df2 = pd.read_csv("/home/msis/ahegde/aml-lab/projects/data/data-2.csv")
```

```python
from sklearn.feature_selection import VarianceThreshold

selector1 = VarianceThreshold(threshold=0.2)
x_selected1 = selector1.fit_transform(df1)
selected_columns1 = df1.columns[selector1.get_support()]

selector2 = VarianceThreshold(threshold=0.2)
x_selected2 = selector2.fit_transform(df2)
selected_columns2 = df2.columns[selector2.get_support()]

print("Selected features (df1):", list(selected_columns1))
print("Selected features (df2):", list(selected_columns2))
```
